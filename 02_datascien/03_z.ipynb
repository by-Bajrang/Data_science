```python
import json

def load_data(filename):
    with open(filename ,"r") as f:  
        return json.load(f)

def find_people_you_know(user_id,data):
    user_friends = {}
    for user in data["users"]:
        user_friends[user['id']] = set(user['friends'])
    if user_id not in user_friends:
        return []

    direct_friends = user_friends[user_id] # if nahi hota tho ye line error thorw kartha 
    suggestions = {}
    for friend in direct_friends:
        for mutul in user_friends[friend]:
            if mutul != user_id and mutul not in direct_friends:
                # count mutul friends
                suggestions[mutul] = suggestions.get(mutul,0) + 1
    return suggestions

# Load data
data = load_data("panda.json")
user_id = 1
recc = find_people_you_know(user_id,data)
print(recc)
    
```
