```python
import json
```

```python
# let write a fxn to load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
        
```

```python
data = load_data("data.json")
```

```python
# write a fxn to define relation of users
def display_users(data):
    print("user and their connections:")
    for user in data['users']:
        print(f"ID:{user['id']}. {user['name']} is friends with ID : {user['friends']}and liked pages are {user['liked_pages']}")
    print("\n pages Information")
    for page in data['pages']:
        print(f"{page['id']} : {page['name']}")

display_users(data)
```
