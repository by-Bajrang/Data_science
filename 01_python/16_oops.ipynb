```python
class Employee:
    company = "HP"

    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
        
    def get_info(self):
        print(f"my name is {self.name} and my salary is {self.salary}")

    @staticmethod  # jab hme self use karna ho tab hm static method use karthe hai
    def time():
        print("time is now")

    @classmethod  # jab app chaho ke class ke atribute use kare samja kya
    def classDetails(cls):
        print(f"the company is {cls.company} ")
        


e = Employee("bajrang",450000)
e.get_info()
e.time()
e.classDetails()

```
