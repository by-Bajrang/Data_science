```python
import numpy as np
import time as time
```

```python
size = 1_000_000
l1 = list(range(size))
l2 = list(range(size))

start = time.time()
add = [ x+y for x,y in zip(l1,l2)]
end = time.time()

print(end-start)
```

```python
size = 1_000_000
l1 = np.array(range(size))
l2 = np.array(range(size))

start = time.time()
add = l1+l2
end = time.time()

print(end-start)
```
