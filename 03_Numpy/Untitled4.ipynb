```python
data = np.array([[10, 20, 30],
                 [15, 25, 35],
                 [20, 30, 40],
                 [25, 35, 45],
                 [30, 40, 50]])
```

```python
mean = data.mean()
```

```python
std = data.std()
```

```python
normalise_data = (data - mean)/std
```
