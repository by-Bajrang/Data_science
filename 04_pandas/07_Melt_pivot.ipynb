```python
import pandas as pd

# sample DataFrame
data = {
    'Name':['Alice','Bob','charlie'],
    'Math':[85,78,92],
    'Science':[90,82,89],
    'English':[88,85,94]
}

df = pd.DataFrame(data)
print(df)
```

```python
meltDf = df.melt(id_vars=["Name"],value_vars=["Math","Science","English"],var_name="Subject",value_name="Score")
```

```python
meltDf
```

```python
meltDf.pivot(index="Name", columns="Subject", values="Score")
```
