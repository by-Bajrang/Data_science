```python
import matplotlib.pyplot as plt
```

```python
# Sample data
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 90, 80, 60, 65, 44, 85, 45]
```

```python
colors = ['red' if score < 50 else 'green' for score in exam_scores]
sizes = [ score*2 for score in exam_scores]

plt.scatter(study_hours,exam_scores,c=colors, s=sizes)
plt.grid(True, color="#6816F7",alpha=0.5)
plt.title("study vs marks")
plt.xlabel("study time")
plt.ylabel("marks")
```

```python

scores = [ 20 , 34 , 34 , 22 , 44, 33 , 22 , 45 ,78 ]

plt.scatter(study_hours,exam_scores,c= scores , cmap = 'viridis') # this scores are show different vlues  
plt.colorbar( label = "score")
plt.grid(True, color="#6816F7",alpha=0.5)
plt.title("study vs marks")
plt.xlabel("study time")
plt.ylabel("marks")
```

```python

scores = [ 20 , 34 , 34 , 22 , 44, 33 , 22 , 45 ,78 ]

plt.scatter(study_hours,exam_scores,c= scores , cmap = 'viridis') # this scores are show different vlues  
plt.colorbar( label = "score")
plt.grid(True, color="#6816F7",alpha=0.5)
plt.title("study vs marks")
plt.xlabel("study time")
plt.ylabel("marks")

# Add label 
for i in range(len(study_hours)):
    plt.annotate(f's{i+1}' , (study_hours[i] + 0.2, exam_scores[i]))
```
