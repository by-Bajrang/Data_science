```python
import matplotlib.pyplot as plt
```

```python
activites = [ "studying" , "playing" , "watching" , "sleepinp"]
time_spent = [ 3,2,2,5]

color = [ "red","#5AB536" , "#1F77DB","#852CE6" ]
# plt.figure(figsize(6,6)
plt.pie(time_spent,labels = activites , colors = color , autopct= "%1.1f%%" )
plt.axis('equal') # use to make perfect circle
```

```python
days = [1, 2, 3, 4, 5, 6, 7]  # Days of the week
studying = [3, 4, 3, 5, 4, 3, 4]
playing = [2, 2, 1, 1, 2, 3, 2]
watching_tv = [2, 1, 2, 2, 1, 1, 1]
sleeping = [5, 5, 6, 5, 6, 5, 5]

label = [ "studying" , "playing" , "TV_watching" , "Sleeping" ]
color = [ "red","#5AB536" , "#1F77DB","#852CE6" ]
plt.figure(figsize=(6,6))
plt.stackplot(days,studying,playing,watching_tv,sleeping,labels = label, colors = color , alpha = 1)
plt.legend(loc="upper right")
plt.title(" weekly Activity Tracker ")
plt.xlabel("days")
plt.ylabel("hours")
plt.grid(True)
```
